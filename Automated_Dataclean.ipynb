```python
from sklearn.impute import SimpleImputer
imputer=SimpleImputer(strategy='median')
imputer1=SimpleImputer(strategy='mean')
imputer2=SimpleImputer(strategy='most_frequent')
imputer3=SimpleImputer(strategy='constant',fill_value=5)
```

```python
df['Heart Rate (BPM)']=imputer.fit_transform(df[['Heart Rate (BPM)']]).ravel()
df['Blood Oxygen Level (%)']=imputer1.fit_transform(df[['Blood Oxygen Level (%)']]).ravel()
df['Activity Level']=imputer2.fit_transform(df[['Activity Level']]).ravel()
df['Stress Level']=imputer3.fit_transform(df[['Stress Level']]).ravel()
df.info( )
```
