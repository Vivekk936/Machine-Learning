```python
data = pd.concat([df, dt], ignore_index=True)
data.shape
```

```python
data['Family Size']=data['SibSp']+data['Parch']+1
```

```python
data=data.drop(columns=['SibSp','Parch'])
```

```python
data.head()
```

```python
sns.barplot(data,x='Family Size',y='Survived')
```

```python
def family_type(n):
    if n>=1 and n<=3:
        return 'Small Family'
    elif n>3 and n<=5:
        return 'Medium Family'
    else:
        return 'Large Family'
    
```

```python
data.shape
```

```python
data['Family Type']=data['Family Size'].apply(family_type)
data['Family Type'].value_counts()
```

```python
data['Family Type']=data['Family Type'].astype('category')
```
