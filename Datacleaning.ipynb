```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('unclean_smartwatch_health_data.csv')
```

```python
df.info()
```

```python
df['Sleep Duration (hours)'] = pd.to_numeric(df['Sleep Duration (hours)'], errors='coerce').astype('float32').round(1)
df
```

```python
df['User ID'] = pd.to_numeric(df['User ID'], errors='coerce').astype('Int32')
df
```

```python
df['Heart Rate (BPM)'] = pd.to_numeric(df['Heart Rate (BPM)'], errors='coerce').astype('float32').round(1)

df
```

```python
df['Blood Oxygen Level (%)'] = pd.to_numeric(df['Blood Oxygen Level (%)'], errors='coerce').astype('float32')

df
```

```python
df['Step Count'] = pd.to_numeric(df['Step Count'], errors='coerce').astype('float32').round(2)

df
```

```python
df['Stress Level'] = pd.to_numeric(df['Stress Level'], errors='coerce').astype('Int32')
df
```
